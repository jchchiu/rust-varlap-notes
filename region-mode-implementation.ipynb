We want to implement a change where we can scan regions by providing a bed file. The previous implementation of regions was to provide a bedfile, and there would be a column in the output that would state whether the variant was within a given region. 

We also want to make it so that ref and alt fields are optional, and that the only required fields are the chromosome and position. This will mainly be implemented for csv/tsv inputs, as VCFs typically require ref/alt fields. Essentially, we just need to change ref/alt to Option<> so that they can take a None value, and refactor the original code to reflect this change. For `Locus/Loci` mode, we need to change the acceptable_variant check so that it can add unknown variants to our variant list; maybe we can set a CLI argument that allows for empty ref/alt. This would most likely be for SNV mode but it can be implemented for both. For unknown variants we will allow for:
- A single acceptable base for ref/alt (A, C, T, G, N) and an empty ref/alt (For SNV mode; should not allow for this implementation in INDEL mode as INDELs require both ref and alt, as the alt gives context for where the INDEL variant starts)
- Both ref/alt are empty (Can be for both SNV/INDEL mode)

To do this we need to make separate CLI commands for `Locus`/`Region` mode; for `Region` mode, we can most likely reuse the SNV feature count part as that also calculates base counts. We can make shared arguments (e.g. reads, threads), and arguments specific to each (e.g. for Locus, we can have the CLI argument for a varaints file and flag for empty ref/alt, while for Region mode we can have the CLI argument for the BED regions file that will create a list of variants).
